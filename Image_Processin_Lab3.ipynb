```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("/content/Sunflower_from_Silesia2.jpg", cv2.IMREAD_GRAYSCALE)
org_img=img
plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")

h,w=img.shape
for i in range(h):
  for j in range(w):
    img[i][j]=np.log(1+img[i][j])

plt.subplot(1,2,2)
plt.title("Log Image")
plt.imshow(img,cmap="gray")
plt.show()
```

```python
log_image=np.log(1+org_img) # directly log transformation of image
plt.title("Log Image")
plt.imshow(log_image,cmap="gray")
```

```python
com_img=255-org_img # directly log transformation of image
plt.title("Complemented Image")
plt.imshow(com_img,cmap="gray")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/Sunflower_from_Silesia2.jpg")

h, w, c = img.shape

border = 50
new_img = np.zeros((h + 2*border, w + 2*border, c),dtype=np.uint8)


for i in range(h):
    for j in range(w):
        new_img[i + border][j + border] = img[i][j]

plt.imshow(cv2.cvtColor(new_img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("/content/Sunflower_from_Silesia2.jpg")

# Add border
border = cv2.copyMakeBorder(
    img,
    50, 50, 50, 50,          # top, bottom, left, right
    cv2.BORDER_CONSTANT,
    value=[100, 100, 100]          # black border
)

# Display
plt.imshow(cv2.cvtColor(border, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/Sunflower_from_Silesia2.jpg")
img2=img.copy()
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
h, w, c = img.shape
for i in range(h):
    for j in range(w):
        img2[i][j] = img2[i][j]-255

plt.subplot(1,2,2)
plt.imshow(img2)
plt.show()
```
