# Assignment 2: RGB to Grayscale Conversion


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('Sunflower_from_Silesia2.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

## Case 1: Mean Average of Three Planes


```python
gray1 = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        gray1[i, j] = np.mean(img[i, j])

plt.figure(figsize=(6, 6))
plt.title('Case 1: Mean Average Grayscale')
plt.imshow(gray1, cmap='gray')
plt.axis('off')
plt.show()
```

## Case 2: User Defined Weights


```python
try:
    wr_in = input('Enter weight for R plane (0 to 1) [Default 0.7]: ')
    wg_in = input('Enter weight for G plane (0 to 1) [Default 0.2]: ')
    wb_in = input('Enter weight for B plane (0 to 1) [Default 0.1]: ')

    w_r = float(wr_in) if wr_in.strip() != '' else 0.7
    w_g = float(wg_in) if wg_in.strip() != '' else 0.2
    w_b = float(wb_in) if wb_in.strip() != '' else 0.1
except (ValueError, EOFError):
    w_r, w_g, w_b = 0.7, 0.2, 0.1

total_w = w_r + w_g + w_b
if not np.isclose(total_w, 1.0):
    w_r, w_g, w_b = w_r / total_w, w_g / total_w, w_b / total_w

print(f'Weights: R={w_r:.3f}, G={w_g:.3f}, B={w_b:.3f}')

gray3 = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        R = img[i, j, 0]
        G = img[i, j, 1]
        B = img[i, j, 2]
        gray3[i, j] = int(round(w_r * R + w_g * G + w_b * B))

plt.figure(figsize=(6, 6))
plt.title(f'Case 2: Weighted (R={w_r:.2f}, G={w_g:.2f}, B={w_b:.2f})')
plt.imshow(gray3, cmap='gray')
plt.axis('off')
plt.show()
```

## Comparison


```python
gray2 = cv2.imread('Sunflower_from_Silesia2.jpg', cv2.IMREAD_GRAYSCALE)

def mse(img1, img2):
    return np.mean(abs(img1.astype(np.float32) - img2.astype(np.float32)) / np.mean(img1))

error1 = mse(gray2, gray1)
error2 = mse(gray2, gray3)

print('Error (Case 1 vs OpenCV):', error1 * 100)
print('Error (Case 2 vs OpenCV):', error2 * 100)

plt.figure(figsize=(14, 5))
plt.subplot(1, 3, 1)
plt.title('Case 1: Mean Average')
plt.imshow(gray1, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title(f'Case 2: Weighted ({w_r:.1f}, {w_g:.1f}, {w_b:.1f})')
plt.imshow(gray3, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('OpenCV Grayscale')
plt.imshow(gray2, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
