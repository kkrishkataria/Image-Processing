```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
imagePath='/content/Sunflower_from_Silesia2.jpg'
image = cv2.imread(imagePath)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.title('Sample Image')
plt.imshow(image_rgb)
plt.show()
```

```python
gray_conversion=cv2.imread(imagePath,cv2.IMREAD_GRAYSCALE)
plt.title('Sample Image (GrayScale Image)')
plt.imshow(gray_conversion,cmap='gray')
plt.show()
```

```python
threshold=np.mean(gray_conversion)
binary=np.where(gray_conversion>threshold,255,0)
plt.title('Sample Image (Binary Image)')
plt.imshow(binary,cmap='gray')
plt.show()
```

```python
total=0.0
pixels=gray_conversion
height, width = pixels.shape[:2]
for y in range(height): # height is no of rows
  for x in range(width): # cols
    total+=pixels[y,x] # row,col
mean_threshhold=total/(width*height)
print(total)
print(mean_threshhold)
binary=np.where(gray_conversion>mean_threshhold,255,0)
plt.title('Sample Image (Binary Image)')
plt.imshow(binary,cmap='gray')
plt.show()
```
