# Assignment 1: Grayscale to Binary Conversion


```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

imagePath = 'Sunflower_from_Silesia2.jpg'
image = cv2.imread(imagePath)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.title('Sample Image')
plt.imshow(image_rgb)
plt.show()
```

```python
gray_conversion = cv2.imread(imagePath, cv2.IMREAD_GRAYSCALE)
plt.title('Sample Image (GrayScale Image)')
plt.imshow(gray_conversion, cmap='gray')
plt.show()
```

## Case 1: Mean Intensity as Threshold


```python
threshold = np.mean(gray_conversion)
binary = np.where(gray_conversion > threshold, 255, 0)
plt.title('Sample Image (Binary Image)')
plt.imshow(binary, cmap='gray')
plt.show()
```

```python
total = 0.0
pixels = gray_conversion
height, width = pixels.shape[:2]
for y in range(height):
    for x in range(width):
        total += pixels[y, x]
mean_threshhold = total / (width * height)
print(total)
print(mean_threshhold)
binary = np.where(gray_conversion > mean_threshhold, 255, 0)
plt.title('Sample Image (Binary Image)')
plt.imshow(binary, cmap='gray')
plt.show()
```

## Case 2: User Input Threshold


```python
try:
    user_input = input('Enter threshold value (0-255) [Default: 120]: ')
    threshold_user = float(user_input) if user_input.strip() != '' else 120.0
except (ValueError, EOFError):
    threshold_user = 120.0

print(f'User Threshold: {threshold_user}')
binary_user = np.where(gray_conversion > threshold_user, 255, 0).astype(np.uint8)

plt.figure(figsize=(6, 6))
plt.title(f'Binary Image (User Threshold = {threshold_user})')
plt.imshow(binary_user, cmap='gray')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.title('Grayscale Image')
plt.imshow(gray_conversion, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title(f'Case 1: Mean Thresh ({mean_threshhold:.1f})')
plt.imshow(binary, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title(f'Case 2: User Thresh ({threshold_user})')
plt.imshow(binary_user, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
