# Assignment 7: Sharpening Filters


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray_img = cv2.imread('Sunflower_from_Silesia2.jpg', cv2.IMREAD_GRAYSCALE)
```

## Question 1: Laplacian Filter


```python
kernel_laplacian = np.array([
    [0,  1, 0],
    [1, -4, 1],
    [0,  1, 0]
], dtype=np.float64)

laplacian_response = cv2.filter2D(gray_img.astype(np.float64), -1, kernel_laplacian)
sharpened_laplacian = np.uint8(np.clip(gray_img.astype(np.float64) - laplacian_response, 0, 255))

kernel_laplacian_diag = np.array([
    [1,  1, 1],
    [1, -8, 1],
    [1,  1, 1]
], dtype=np.float64)
lap_diag_response = cv2.filter2D(gray_img.astype(np.float64), -1, kernel_laplacian_diag)
sharpened_lap_diag = np.uint8(np.clip(gray_img.astype(np.float64) - lap_diag_response, 0, 255))

plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.title('Original Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title('Laplacian Edge Map')
plt.imshow(laplacian_response, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title('Sharpened (Laplacian 4-conn)')
plt.imshow(sharpened_laplacian, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title('Sharpened (Laplacian 8-conn)')
plt.imshow(sharpened_lap_diag, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Question 2: Sobel and Canny Edge Detectors


```python
sobel_x = cv2.Sobel(gray_img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray_img, cv2.CV_64F, 0, 1, ksize=3)
sobel_mag = np.sqrt(sobel_x**2 + sobel_y**2)
sobel_norm = np.uint8(np.clip((sobel_mag / np.max(sobel_mag)) * 255.0, 0, 255))
sobel_sharpened = np.uint8(np.clip(gray_img.astype(np.float64) + 0.5 * sobel_mag, 0, 255))

canny_edges = cv2.Canny(gray_img, threshold1=80, threshold2=180)

plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.title('Original Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title('Sobel |Gx|')
plt.imshow(np.abs(sobel_x), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title('Sobel |Gy|')
plt.imshow(np.abs(sobel_y), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title('Sobel Magnitude')
plt.imshow(sobel_norm, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title('Sobel Sharpened')
plt.imshow(sobel_sharpened, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title('Canny Edge Detector')
plt.imshow(canny_edges, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Question 3: Unsharp Masking


```python
blurred = cv2.GaussianBlur(gray_img, (5, 5), sigmaX=3.0, sigmaY=3.0)
mask = gray_img.astype(np.float64) - blurred.astype(np.float64)

k1 = 1.0
sharpened_unsharp = np.uint8(np.clip(gray_img.astype(np.float64) + k1 * mask, 0, 255))

k2 = 2.0
sharpened_highboost = np.uint8(np.clip(gray_img.astype(np.float64) + k2 * mask, 0, 255))

plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.title('Original Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title('Gaussian Blurred (5x5, \u03c3=3)')
plt.imshow(blurred, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title('Unsharp Mask')
plt.imshow(mask + 128, cmap='gray', vmin=0, vmax=255)
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title('Unsharp Masking (k = 1)')
plt.imshow(sharpened_unsharp, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title('High-Boost Filtering (k = 2)')
plt.imshow(sharpened_highboost, cmap='gray')
plt.axis('off')

crop_y, crop_x = 200, 200
crop_h, crop_w = 120, 120
plt.subplot(2, 3, 6)
plt.title('Detail (k=2 Sharpened)')
plt.imshow(sharpened_highboost[crop_y:crop_y+crop_h, crop_x:crop_x+crop_w], cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
