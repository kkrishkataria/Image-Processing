# Assignment 3


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray_img = cv2.imread('Sunflower_from_Silesia2.jpg', cv2.IMREAD_GRAYSCALE)
threshold_val = np.mean(gray_img)
bin_img = np.where(gray_img > threshold_val, 255, 0).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Original Grayscale Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Original Binary Image')
plt.imshow(bin_img, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Question 1: Border Padding


### Grayscale Image


```python
try:
    w_in = input('Enter border width in pixels [Default 40]: ')
    c_in = input('Enter border color/intensity (0-255) [Default 180]: ')
    pad_w_gray = int(w_in) if w_in.strip() != '' else 40
    pad_c_gray = int(c_in) if c_in.strip() != '' else 180
except (ValueError, EOFError):
    pad_w_gray = 40
    pad_c_gray = 180

h, w = gray_img.shape
padded_gray = np.full((h + 2 * pad_w_gray, w + 2 * pad_w_gray), pad_c_gray, dtype=np.uint8)
padded_gray[pad_w_gray:pad_w_gray + h, pad_w_gray:pad_w_gray + w] = gray_img

plt.figure(figsize=(11, 5))
plt.subplot(1, 2, 1)
plt.title('Grayscale Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title(f'Padded Grayscale (Width={pad_w_gray}, Color={pad_c_gray})')
plt.imshow(padded_gray, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Binary Image


```python
try:
    w_bin_in = input('Enter border width in pixels [Default 40]: ')
    c_bin_in = input('Enter border color (0 or 255) [Default 255]: ')
    pad_w_bin = int(w_bin_in) if w_bin_in.strip() != '' else 40
    pad_c_bin = int(c_bin_in) if c_bin_in.strip() != '' else 255
except (ValueError, EOFError):
    pad_w_bin = 40
    pad_c_bin = 255

h, w = bin_img.shape
padded_bin = np.full((h + 2 * pad_w_bin, w + 2 * pad_w_bin), pad_c_bin, dtype=np.uint8)
padded_bin[pad_w_bin:pad_w_bin + h, pad_w_bin:pad_w_bin + w] = bin_img

plt.figure(figsize=(11, 5))
plt.subplot(1, 2, 1)
plt.title('Binary Image')
plt.imshow(bin_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title(f'Padded Binary (Width={pad_w_bin}, Color={pad_c_bin})')
plt.imshow(padded_bin, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## Question 2: Image Complement


### Grayscale Image


```python
h, w = gray_img.shape
comp_gray = np.zeros((h, w), dtype=np.uint8)
for i in range(h):
    for j in range(w):
        comp_gray[i, j] = 255 - gray_img[i, j]

plt.figure(figsize=(11, 5))
plt.subplot(1, 2, 1)
plt.title('Original Grayscale Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Grayscale Complement (255 - r)')
plt.imshow(comp_gray, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Binary Image


```python
comp_bin = np.zeros((h, w), dtype=np.uint8)
for i in range(h):
    for j in range(w):
        comp_bin[i, j] = 255 - bin_img[i, j]

plt.figure(figsize=(11, 5))
plt.subplot(1, 2, 1)
plt.title('Original Binary Image')
plt.imshow(bin_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Binary Complement')
plt.imshow(comp_bin, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```
