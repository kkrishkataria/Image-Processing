```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray_img = cv2.imread('Sunflower_from_Silesia2.jpg', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(6, 6))
plt.title('Original Grayscale Image')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')
plt.show()
```

## Question 1: Log Transformation (c=1)


```python
c = 1.0
log_raw = c * np.log(1.0 + gray_img.astype(np.float64))
c_scaled = 255.0 / np.log(1.0 + np.max(gray_img))
log_scaled = np.uint8(c_scaled * np.log(1.0 + gray_img.astype(np.float64)))

plt.figure(figsize=(14, 5))
plt.subplot(1, 3, 1)
plt.title('Original Grayscale')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('Log Transform (c=1, Float)')
plt.imshow(log_raw, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Log Transform (Scaled [0, 255])')
plt.imshow(log_scaled, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Question 2: Gamma Transformation (c=1)


```python
try:
    g_in = input('Enter gamma value [Default 0.5]: ')
    gamma = float(g_in) if g_in.strip() != '' else 0.5
except (ValueError, EOFError):
    gamma = 0.5

norm_img = gray_img.astype(np.float64) / 255.0
c = 1.0
gamma_out = c * np.power(norm_img, gamma)
gamma_img = np.uint8(np.clip(gamma_out * 255.0, 0, 255))

g_low = 0.4
g_high = 2.2
img_low = np.uint8(np.clip(np.power(norm_img, g_low) * 255.0, 0, 255))
img_high = np.uint8(np.clip(np.power(norm_img, g_high) * 255.0, 0, 255))

plt.figure(figsize=(15, 5))
plt.subplot(1, 4, 1)
plt.title('Original Grayscale')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title(f'Gamma = {gamma}')
plt.imshow(gamma_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title(f'Gamma = {g_low}')
plt.imshow(img_low, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title(f'Gamma = {g_high}')
plt.imshow(img_high, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Question 3: Contrast Stretching


```python
try:
    rmin_in = input('Enter rmin [Default 80]: ')
    rmax_in = input('Enter rmax [Default 120]: ')
    smin_in = input('Enter smin [Default 50]: ')
    smax_in = input('Enter smax [Default 150]: ')

    rmin = int(rmin_in) if rmin_in.strip() != '' else 80
    rmax = int(rmax_in) if rmax_in.strip() != '' else 120
    smin = int(smin_in) if smin_in.strip() != '' else 50
    smax = int(smax_in) if smax_in.strip() != '' else 150
except (ValueError, EOFError):
    rmin, rmax, smin, smax = 80, 120, 50, 150

img_f = gray_img.astype(np.float64)
stretched = np.zeros_like(img_f)

if rmin > 0:
    mask1 = img_f < rmin
    stretched[mask1] = (smin / rmin) * img_f[mask1]

mask2 = (img_f >= rmin) & (img_f <= rmax)
if rmax > rmin:
    stretched[mask2] = ((smax - smin) / (rmax - rmin)) * (img_f[mask2] - rmin) + smin
else:
    stretched[mask2] = smin

if rmax < 255:
    mask3 = img_f > rmax
    stretched[mask3] = ((255.0 - smax) / (255.0 - rmax)) * (img_f[mask3] - rmax) + smax

stretched_img = np.uint8(np.clip(stretched, 0, 255))

sliced_preserve = gray_img.copy()
sliced_preserve[mask2] = np.uint8(np.clip(((smax - smin) / (rmax - rmin)) * (img_f[mask2] - rmin) + smin, 0, 255))

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.title('Original Grayscale')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title(f'Contrast Stretched [{rmin}, {rmax}] -> [{smin}, {smax}]')
plt.imshow(stretched_img, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Intensity Level Sliced')
plt.imshow(sliced_preserve, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
