```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('Sunflower_from_Silesia2.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
```

```python
grey_img=cv2.cvtColor(img,cv2.COLOR_RGBA2GRAY)
plt.imshow(grey_img,cmap='gray')
```

```python
h,w=grey_img.shape
log_img=grey_img.copy()
for i in range(h):
  for j in range(w):
    log_img[i,j]=np.log(1+grey_img[i,j])

plt.imshow(log_img,cmap='gray')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("Sunflower_from_Silesia2.jpg", 0)

gamma = float(input("Enter gamma value: ")) # bw 1 and 0 bright greater than 1
# darker image will be formed


gamma_img = np.power(img, gamma)


plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gamma_img, cmap='gray')
plt.title("Gamma Transformed")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("Sunflower_from_Silesia2.jpg",cv2.IMREAD_COLOR_RGB)

rmin = int(input("Enter input minimum intensity: "))
rmax = int(input("Enter input maximum intensity: "))

smin = int(input("Enter output minimum intensity: "))
smax = int(input("Enter output maximum intensity: "))
output=img.copy()

# Contrast stretching
mask = (img >= rmin) & (img <= rmax)

output[mask] = ((img[mask] - rmin) * (smax - smin)) / (rmax - rmin) + smin


# Display
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(output, cmap='gray')
plt.title("Contrast Stretched")
plt.axis('off')

plt.show()
```
